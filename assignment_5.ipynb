```python
#to implement z-score normalization , min-max normalization & decimal scaling normalization on 10,20,30,40,50
```

```python
import numpy as np
data = np.array([10, 20, 30, 40, 50], dtype=float)

mean = np.mean(data)

sd = np.std(data)

z_score = (data - mean) / sd

min_value = np.min(data)
max_value = np.max(data)

min_max = (data - min_value) / (max_value - min_value)

decimal_scaling = data / 100

print("Original Data:")
print(data)

print("\nZ-Score Normalization:")
print(z_score)

print("\nMin-Max Normalization:")
print(min_max)

print("\nDecimal Scaling Normalization:")
print(decimal_scaling)
```

```python
#to implement the decimal scaling normalization on a dataset(using kaggle dataset)
```

```python
import pandas as pd
import numpy as np

data = pd.read_csv("Iris.csv")

x = data["PetalLengthCm"]

max_value = x.abs().max()

j = len(str(int(max_value)))

normalized = x / (10 ** j)

print("Original Values:")
print(x.head(10))

print("\nMaximum Value:", max_value)
print("Value of j:", j)

print("\nDecimal Scaling Normalized Values:")
print(normalized.head(10))
```

```python
import pandas as pd

data = pd.read_csv("Iris.csv")

x = data["SepalWidthCm"]

minimum = x.min()
maximum = x.max()

normalized = (x - minimum) / (maximum - minimum)

print("Original Values:")
print(x.head(10))

print("\nMinimum Value:", minimum)
print("Maximum Value:", maximum)

print("\nMin-Max Normalized Values:")
print(normalized.head(10))
```

```python
import pandas as pd
import numpy as np

data = pd.read_csv("Iris.csv")

x = data["PetalWidthCm"]

mean = np.mean(x)

variance = np.var(x)

sd = np.std(x)

z_score = (x - mean) / sd

print("Mean =", mean)
print("Variance =", variance)
print("Standard Deviation =", sd)

print("\nOriginal Values:")
print(x.head(10))

print("\nZ-Score Normalized Values:")
print(z_score.head(10))
```
