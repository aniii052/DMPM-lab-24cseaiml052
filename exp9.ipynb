```python
#findout the frequent item sets-> consider the following transaction database:
#transaction id: T1,T2,T3,T4,T5
#Items:{bread,milk},{bread,diaper,beer,eggs},{milk,diaper,beer,coke},{bread,milk,diaper,beer},{bread,milk,diaper,coke}
```

```python
from itertools import combinations

T = [
    {"bread","milk"},
    {"bread","diaper","beer","eggs"},
    {"milk","diaper","beer","coke"},
    {"bread","milk","diaper","beer"},
    {"bread","milk","diaper","coke"}
]

min_sup = 2

items = set().union(*T)

for k in range(1, 4):
    for c in combinations(items, k):
        s = sum(set(c) <= t for t in T)
        if s >= min_sup:
            print(c, ":", s)
```

```python
#one hundred people were interviewed outside a chocolate shop to find out which flavour of chocolate cream they prefered. the results are given in the table below classified by gender.
#gender strawberry coffee orange vanila
#male 23 18 8 9
#female 15 6 12 10
#we want to determine whether the flavour of chocolate cream is independent of gender:
#a)state the null hypothesis & the alternative hypothesis
#b)calculate the df
#c) create the expected frequency table for the data
```

```python
import numpy as np
from scipy.stats import chi2_contingency

data = np.array([
    [23, 18, 8, 9],
    [15, 6, 12, 10]
])

print("H0: Chocolate flavour is independent of gender")
print("H1: Chocolate flavour is not independent of gender")

df = (data.shape[0] - 1) * (data.shape[1] - 1)
print("Degrees of freedom:", df)

chi2, p, df, expected = chi2_contingency(data)

print("Expected Frequency Table:")
print(expected)
```
