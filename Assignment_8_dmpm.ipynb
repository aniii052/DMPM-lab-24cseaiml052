```python
#calculate & analyze the correlation coefficient & covariance between 2 variables using given data
```

```python
X = [10, 20, 30, 40, 50]
Y = [20, 25, 35, 45, 55]

n = len(X)

mean_x = sum(X) / n
mean_y = sum(Y) / n

sum_xy = 0
sum_x2 = 0
sum_y2 = 0

for i in range(n):
    sum_xy = sum_xy + (X[i] - mean_x) * (Y[i] - mean_y)
    sum_x2 = sum_x2 + (X[i] - mean_x) ** 2
    sum_y2 = sum_y2 + (Y[i] - mean_y) ** 2

covariance = sum_xy / n

correlation = sum_xy / ((sum_x2 * sum_y2) ** 0.5)

print("Mean of X =", mean_x)
print("Mean of Y =", mean_y)
print("Covariance =", covariance)
print("Correlation coefficient =", correlation)

if correlation > 0:
    print("There is a positive correlation.")
elif correlation < 0:
    print("There is a negative correlation.")
else:
    print("There is no correlation.")
```
